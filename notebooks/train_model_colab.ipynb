```python
import base64, json

export = {}
for path in ["results/models/best_model.joblib", "results/metrics/model_comparison.json"]:
    with open(path, "rb") as f:
        export[path] = base64.b64encode(f.read()).decode()
print("BDA_EXPORT_BEGIN")
print(json.dumps(export))
print("BDA_EXPORT_END")
```
